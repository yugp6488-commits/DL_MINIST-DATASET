```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader


transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,))
])


train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
val_dataset = datasets.MNIST(root='./data', train=False, download=True, transform=transform)


train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=32, shuffle=False)


dataiter = iter(train_loader)
images, labels = next(dataiter)
print("Batch image shape:", images.shape) # Output: torch.Size([32, 1, 28, 28])
```

```python
import tensorflow as tf

rnn = tf.keras.models.Sequential()


rnn.add(tf.keras.layers.Input(shape=(28, 28)))

rnn.add(tf.keras.layers.LSTM(128))

rnn.add(tf.keras.layers.Dense(10, activation='softmax'))

# 5. Compile the architecture
rnn.compile(optimizer='adam',
            loss='sparse_categorical_crossentropy',
            metrics=['accuracy'])

rnn.summary()
```

```python
import tensorflow as tf

(x_train, y_train), (x_val, y_val) = tf.keras.datasets.mnist.load_data()


x_train = x_train / 255.0
x_val = x_val / 255.0

print("Starting RNN training...")

history_rnn = rnn.fit(x_train,
                      y_train,
                      epochs=5,
                      validation_data=(x_val, y_val))
```
